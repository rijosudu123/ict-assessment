#Merge all four datasets into one DataFrame.
Hint:

● Use the common column (country)

● Choose an appropriate join method.

```python
merge1 = pd.merge(population,gdp, on = "country",how = "inner")
merge2 = pd.merge(merge1,interuser, on = "country", how = "inner")
merge3 = pd.merge(merge2,literacyrate, on = "country", how = "inner")
merge3
```

#Task 9
Create the following calculated columns:

1. Internet Penetration (%)

2. GDP Per Capita

```python
merge3["internet_penetration (%)"] = (merge3['internet_users']/merge3['population'])*100
merge3["GDP per capital"] = merge3['GDP']/merge3['population']
merge3
```

#Task 10
Create a SQLite database. Store the final merged dataset into a table named

‘country_statistics’

Hint:

● Use the sqlite3 library.

● Load the DataFrame using pandas.

```python
import sqlite3

try :
  conn = sqlite3.connect("country_statistics.db")
  merge3.to_sql("country_statistics",conn,if_exists="replace",index=False)
  print("Database created and connected successfully")
except sqlite3.Error as e :
  print("Error connecting to database",e)
finally :
  conn.close()
```

```python
import sqlite3

conn = sqlite3.connect("country_statistics.db")
cursor = conn.cursor()
query = """
SELECT country, GDP
FROM country_statistics
ORDER BY GDP DESC
LIMIT 5
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT AVG(literacy_rate) AS average_literacy_rate
FROM country_statistics
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT country, literacy_rate
FROM country_statistics
WHERE literacy_rate > 90
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT country, [Internet_penetration (%)]
FROM country_statistics
ORDER BY [Internet_penetration (%)] DESC
LIMIT 5;
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT SUM(population) AS total_population
FROM country_statistics
"""

pd.read_sql(query, conn)
```

```python
query = """
SELECT country,
(GDP*1.0/population) AS GDP_per_capita
FROM country_statistics
ORDER BY GDP_per_capita DESC
LIMIT 1
"""

pd.read_sql(query, conn)
```

```python
query = "SELECT * FROM country_statistics"

df = pd.read_sql(query, conn)

df.to_csv(
    "final_merged_dataset.csv",
    index=False
)

print("Exported")
```

```python
query = """
SELECT country,
GDP,
population,
(GDP*1.0/population) AS GDP_per_capita
FROM country_statistics
ORDER BY GDP_per_capita DESC
LIMIT 10
"""

pd.read_sql(query, conn)
```
