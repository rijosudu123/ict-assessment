 Create a new column Profitability:
a) Revenue divided by Duration
b) Find top 10 most profitable movies

```python
data_new['profitability'] = data_new['Revenue (Millions)']/data_new['Runtime (Minutes)']
data_new
```

```python
profit=data_new.sort_values(by = 'profitability',ascending=False).head(10)
profit[['movie title','profitability']]
```

```python
def quality_check(Rating):
  if Rating < 6 :
     return 'poor'
  elif Rating < 7.5:
    return 'Average'
  else:
     return 'Excellent'
data_new['quality'] = data_new['Rating'].apply(quality_check)
data_new
```

```python
data_new['High_Score'] = data_new['Rating'] * 10 + data_new['Revenue (Millions)']
data_new
```
