```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
train = pd.read_csv("/content/train.csv")
train
```

```python
train.info()
```

```python
train.describe()
```

```python
train.isna().sum()  #finding the missing value
```

```python
train["peso"].skew() # check the veriable is skewed or not
```

```python
train["peso"] = train["peso"].fillna(train["peso"].mean()) #fill the missing value using mean
```

```python
train.isna().sum() #again check the data
```

```python
train.duplicated()  #finding the duplicates
```

```python
train = train.drop(columns = ['id'])  #drop id
```

```python
train.corr(numeric_only = True)
```

```python
train = train.drop(columns = ['mediaEnvergadura'])
```

```python
num_col = train.select_dtypes(include=["float64"]).columns
for col in num_col:
    plt.figure(figsize=(6,5))
    sns.boxplot(train[col])
    plt.title(f'outliers in {col}')
    plt.show()
```

```python
for col in num_col:
  q1 = train[col].quantile(0.25)
  q3 = train[col].quantile(0.75)
  iqr = q3-q1
  lower_bound  = q1-1.5*iqr
  upper_bound =  q3+1.5*iqr
  train[col] = train[col].clip(lower=lower_bound,upper=upper_bound)
```

```python
for col in num_col:
  sns.histplot(data=train,x=col,hue = 'sexo',kde=True)
  plt.title(f'releation{col}')
  plt.show()
```

```python
from sklearn.preprocessing import LabelEncoder
label_enc = LabelEncoder()
train['sexo']=label_enc.fit_transform(train['sexo'])
```

```python
train
```

```python
x = train.drop(columns = ['sexo'])
y = train['sexo']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score,classification_report
```

```python
model = LogisticRegression()
model.fit(x_train,y_train)
y_pred = model.predict(x_test)
print("logistic regression:", accuracy_score(y_test,y_pred))
print("classification matrix:",classification_report(y_test,y_pred))
```

```python
model_dt = DecisionTreeClassifier(random_state=42)
model_dt.fit(x_train, y_train)
y_pred_dt = model_dt.predict(x_test)
print("Decision Tree Accuracy:", accuracy_score(y_test, y_pred_dt))
print("classification matrix:",classification_report(y_test,y_pred_dt))
```

```python
model_rf = RandomForestClassifier(random_state=42)
model_rf.fit(x_train, y_train)
y_pred_rf = model_rf.predict(x_test)
print("Random Forest Accuracy:", accuracy_score(y_test, y_pred_rf))
print("classification matrix:",classification_report(y_test,y_pred_rf))
```

```python
model_svc = SVC()
model_svc.fit(x_train, y_train)
y_pred_svc = model_svc.predict(x_test)
print("SVC Accuracy:", accuracy_score(y_test, y_pred_svc))
print("classification matrix:",classification_report(y_test,y_pred_svc))
```

```python
model_gnb = GaussianNB()
model_gnb.fit(x_train, y_train)
y_pred_gnb = model_gnb.predict(x_test)
print("Gaussian Naive Bayes Accuracy:", accuracy_score(y_test, y_pred_gnb))
print("classification matrix:",classification_report(y_test,y_pred_gnb))
```

```python
model_knn = KNeighborsClassifier()
model_knn.fit(x_train, y_train)
y_pred_knn = model_knn.predict(x_test)
print("KNN Accuracy:", accuracy_score(y_test, y_pred_knn))
print("classification matrix:",classification_report(y_test,y_pred_knn))
```

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    'n_neighbors': [3, 5, 7, 9, 11],
    'weights': ['uniform', 'distance'],
    'metric': ['euclidean', 'manhattan']
}

grid_search = GridSearchCV(KNeighborsClassifier(), param_grid, cv=5, scoring='accuracy')
grid_search.fit(x_train, y_train)

best_knn = grid_search.best_estimator_
print("Best Parameters:", grid_search.best_params_)
print("Tuned Accuracy:", best_knn.score(x_test, y_test))
```
